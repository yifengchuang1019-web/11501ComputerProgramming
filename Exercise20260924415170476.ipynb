## 第五部分：Python 整數與浮點數

回答：

1. 為什麼 Python 能表示如此大的整數？Python 能表示這些數，因為 int 是任意精度整數，不限於 CPU 的 32 或 64 位元整數。
2. Python 的 `int` 是否具有固定的位元數限制？int 沒有固定的位元數限制。
3. 什麼是 Arbitrary Precision（任意精度）？Arbitrary Precision（任意精度）：整數所需位元越多，實作就配置越多儲存區塊。

3. 什麼是浮點數的精度（precision）？精度（precision）：能保留多少有效數字／相鄰可區分值有多細。float 通常是 53 個二進位有效位元，約 15–17 位十進位有效數字。
4. 什麼是浮點數的範圍（range）？範圍（range），可表示數值的大小跨度。
5. 精度與範圍有何不同？精度關心「相近數值能否分辨」；範圍關心「最大與最小量級能到哪裡」。兩者由不同位元欄位主導：有效數字影響精度，指數影響範圍。

## 第六部分：AI 模型中的數值表示

人工智慧模型需要儲存大量參數（Parameters）與權重（Weights）。

請查閱資料並回答：

1. AI 模型中的權重通常使用什麼型別儲存？權重通常存成張量中的浮點格式：FP32、FP16 或 BF16；現代大型模型很常採混合精度。
2. 什麼是 FP64？FP64：64-bit IEEE 754 binary64，1 個符號位元、11 個指數位元、52 個儲存的尾數位元（有效精度為 53 bits）。
3. 什麼是 FP32？FP32：32-bit IEEE 754 binary32，1 + 8 + 23 bits。
4. 什麼是 FP16？FP16：16-bit IEEE 754 binary16，1 + 5 + 10 bits。
5. 什麼是 BF16？BF16：16-bit Brain Floating Point，1 + 8 + 7 bits；它保留與 FP32 相近的指數範圍，但有效精度較低。
6. 為什麼大型 AI 模型不直接使用 Python 的 `int` 作為權重？不用 Python int 當權重，因為權重通常是連續的近似實數，需要小數；而 Python 整數是每個值都是獨立、高額外負擔的 Python 物件，不適合放入數十億個 GPU 張量元素，也不吻合 GPU 的矩陣乘法硬體格式。

完成下表。

| 表示方式 | 位元數 | 是否有誤差 | 主要用途 |
|----------|--------|------------|----------|
| int | 可變 | 整數值本身沒有表示誤差 | 計數、索引、精確整數與大數運算 |
| FP64 | 64 | 是 | 高精度科學、工程與統計計算 |
| FP32 | 32 | 是 | 一般數值運算、傳統深度學習、部分累積值 |
| FP16 | 16 | 是 | 高吞吐 AI 訓練與推論 |
| BF16 | 16 | 是 | 大型模型訓練與推論，尤其重視範圍時 |

並回答：

1. 哪種表示方式最精確？在此表中，對一般實數而言 FP64 最精確；但對整數而言，int 對可容納的整數是完全精確的。
2. 哪種表示方式最節省記憶體？FP16 與 BF16 都最省記憶體，每個元素 2 bytes。
3. 哪種表示方式最常見於 AI 訓練？AI 訓練目前常見的是混合精度：FP16 或 BF16 做大量矩陣運算，搭配部分 FP32（例如累積、最佳化器狀態或主權重）；選擇取決於 GPU/TPU 與模型。
4. 哪種表示方式最常見於 AI 推論（Inference）？推論在這四種浮點格式中常見 FP16 或 BF16；實務上也常使用 INT8、FP8 等量化格式以進一步節省記憶體與提升吞吐量。

## 第八部分：綜合思考

### 思考題一

Python 官方文件指出：

```python
int
```

具有 Unlimited Precision（無限制精度）。

請回答：

1. Unlimited Precision 是什麼意思？指整數的表示長度可隨數值增加，不預先固定為 32 或 64 bits。
2. 它是否代表可以儲存無限大的整數？不代表能儲存「真正無限大」的整數；電腦仍有有限記憶體與時間。
3. CPython 如何實現這項特性？CPython 的 PyLongObject 以多個固定寬度的數字區塊組成整數。數字變大時，需要更多區塊；算術結果會建立足以容納結果的新物件。

### 思考題二

比較：

```python
int
```

與

```python
float
```

回答：

1. 為什麼 `int` 幾乎沒有固定上限？int 可增加更多區塊，因此沒有固定的數學上限。
2. 為什麼 `float` 卻存在範圍限制？float 使用固定長度的 IEEE 754 格式，指數欄位位數有限，所以可表示的最大量級有限。
3. 為什麼 `float` 可能產生誤差？float 的有效數字位元有限，而許多數值（例如 0.1）在二進位是無限循環小數，必須四捨五入。

### 思考題三

比較下列兩種設計理念：

#### Python 的整數

```text
Arbitrary Precision
正確性優先
```

#### AI 模型常用的 FP16、BF16

```text
Limited Precision
效率優先
```

請說明：

1. 兩種設計理念的差異。差異：int 是正確性優先的任意精度整數；FP16／BF16 是效率優先的固定精度浮點數。
2. 各自的優點與缺點。優缺點：int 精確但大數較耗記憶體與時間；FP16／BF16 快、省記憶體但有誤差與範圍限制。
3. 為什麼大型語言模型通常選擇較低精度的表示方式？LLM 選低精度：可節省權重與中間張量的記憶體、降低資料搬運量，並使用 GPU／TPU 的高吞吐低精度矩陣硬體。
4. 如果所有 AI 模型都改用 FP64，可能會產生哪些問題？全用 FP64：記憶體約為 FP16／BF16 的四倍，頻寬需求與成本升高，batch size 變小，而且多數硬體上的運算速度較慢。

### 挑戰題

請用自己的話解釋：

```python
x = 10**1000
```

為什麼 Python 可以正常運作；

而大型 AI 模型卻通常選擇使用：

```text
FP32
FP16
BF16
```

而不是任意精度整數或高精度浮點數？

請從：

- 記憶體使用量
- 計算速度
- 數值精度
- GPU 運算效率

等角度進行討論。

x = 10**1000 能正常運作，是因為它只是一個整數，而 CPython 讓 int 依需要增加儲存區塊；它的代價是更多記憶體與更慢的大數運算，但 1,000 位十進位數其實仍很小。
大型 AI 模型則要同時保存、搬運並重複運算數十億甚至更多權重。若每個權重都用任意精度整數，不但不能自然表示小數權重，每個元素的大小還不固定、物件額外負擔高，也無法直接使用 GPU 的密集矩陣乘法硬體。若全面改為 FP64，雖可減少部分數值誤差，但每個權重需 8 bytes，記憶體頻寬與儲存成本會顯著上升，GPU 吞吐量往往下降。
因此，AI 通常接受 FP32、FP16、BF16 的有限誤差：它們以固定 2 或 4 bytes 的規格儲存，每個張量元素排列緊密，能由 GPU/TPU 平行矩陣運算單元高效率處理。這是「在模型品質可接受的前提下，以較小誤差換取大量記憶體、速度與能源效益」的工程取捨。
