1. 0.1的二進位表示法為何? 使用markdown語法，列出計算過程和理由，並描述看到的現象。
小數部分連續乘以 2，每次取整數部分當作一個位元，剩下的小數繼續乘 2。
| 計算 | 整數位 | 剩餘小數 |
|---|---:|---:|
| \(0.1 \times 2 = 0.2\) | 0 | 0.2 |
| \(0.2 \times 2 = 0.4\) | 0 | 0.4 |
| \(0.4 \times 2 = 0.8\) | 0 | 0.8 |
| \(0.8 \times 2 = 1.6\) | 1 | 0.6 |
| \(0.6 \times 2 = 1.2\) | 1 | 0.2 |
看到的現象
十進位的 0.1 是有限小數，但轉成二進位後會變成無限循環小數，循環部分是 0011。因此，有限精度的二進位浮點數無法精確儲存 0.1，只能儲存非常接近它的值。

2. 如果是無限多位，Python如何解決？提示：IEEE 754、四捨六入五成雙
Python 的一般 float 使用 IEEE 754 雙精度格式，能儲存的位數有限，所以無法完整保存像 \(0.1_{10}=0.000110011\ldots_2\) 這種無限循環的二進位小數。
Python 會把它捨入成最接近的可表示浮點數。若剛好介於兩個可表示值的正中間，則選擇尾數最低位為偶數的那個值，這稱為「最接近值、五成雙」（round to nearest, ties to even）。
因此，Python 並沒有保存無限多位，而是保存有限精度的近似值。

3. Python如何知道浮點數的精度？請問執行以下程式碼:
```python
import sys
print(sys.float_info)
```
並說明輸出結果的意思。
Python 透過 sys.float_info 取得目前執行環境中 float 格式的精度資訊。它反映底層浮點數格式（通常是 IEEE 754 雙精度），例如 mant_dig 表示二進位有效位數，dig 表示十進位有效數字的精度。
max、min：最大有限值，以及最小的正規格化正值。
max_exp、min_exp：二進位指數範圍；max_10_exp、min_10_exp 是十進位指數範圍。
mant_dig=53：尾數有 53 個二進位有效位元。
dig=15：約可保證 15 位十進位有效數字的精度。
epsilon：1.0 和比 1.0 大的下一個可表示浮點數之間的距離。
radix=2：浮點數以 2 為底。
rounds=1：通常代表運算採用就近捨入。
