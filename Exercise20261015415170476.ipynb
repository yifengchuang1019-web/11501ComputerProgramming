習題一：使用random模組，產生出一個介於0到1之間的小數$r$，使用while迴圈，計算出$r$的前二十位二進位表示$r = 0.b_1b_2b_3\cdots {b_{20}}_2$。輸出格式如附圖Exercise20261015.png

````python
import random

r = random.random()  # 產生 0 ≤ r < 1 的小數
x = r                # 用 x 計算，保留 r 作為原始值
bits = ""
count = 0

while count < 20:
    x *= 2
    if x >= 1:
        bits += "1"
        x -= 1
    else:
        bits += "0"
    count += 1

print(f"假如 r = {r}")
print(f"r ≈ 0.{bits}₂")
````

Python 的 `int` 型別採用 Arbitrary-Precision Integer（任意精度整數）設計，能夠精確表示任意大小的整數，其實作方式可參考 CPython 的 `PyLongObject`。與 C/C++ 的 `int` 或 `long long` 不同，Python 的整數不受固定 32 位元或64 位元限制，而是會依據數值大小自動配置更多記憶體，因此只要記憶體足夠，便能表示極大的整數。例如：
```python
n = 10**68
print(n)
```
上述程式可正常執行，表示 Python 能夠精確儲存$10^{68}$，即中文數字中的「無量大數」。


可利用下列程式觀察不同大小整數所需的位元數（bit）與記憶體空間：

```python
import sys
n = 10**68
print(f"bit_length = {n.bit_length()}")
print(f"memory = {sys.getsizeof(n)} bytes")
```

並查閱官方文件，了解：

- `bit_length()`
- `sys.getsizeof()`
- `PyLongObject`

三者之間的關係。

習題二：輸入一個正整數 \(n\)，其中$1 \le n \le 10^{68}$，請將 \(n\) 轉換為中文大寫數字表示法。需支援下列數字單位：萬、億、兆、京、垓、秭、穰、溝、澗、正、載、極、恆河沙、阿僧祇、那由他、不可思議、無量大數。比如輸入31415926，輸出參仟壹佰肆拾壹萬伍仟玖佰貳拾陸。

提示：思考利用字典，建立「數字」與「中文大寫數字」的對應關係。

````python
digits = "零壹貳參肆伍陸柒捌玖"
small_units = ["", "拾", "佰", "仟"]
large_units = [
    "", "萬", "億", "兆", "京", "垓", "秭", "穰", "溝",
    "澗", "正", "載", "極", "恆河沙", "阿僧祇", "那由他",
    "不可思議", "無量大數"
]


def convert_group(group):
    """把 0 到 9999 轉成中文大寫數字。"""
    result = ""
    zero_pending = False

    position = 3
    while position >= 0:
        divisor = 10 ** position
        digit = group // divisor
        group %= divisor

        if digit == 0:
            if result:
                zero_pending = True
        else:
            if zero_pending:
                result += "零"
            result += digits[digit] + small_units[position]
            zero_pending = False

        position -= 1

    return result


def convert_number(n):
    if not 1 <= n <= 10**68:
        return "輸入範圍需為 1 到 10^68"

    groups = []
    while n > 0:
        groups.append(n % 10000)
        n //= 10000

    result = ""
    zero_pending = False
    index = len(groups) - 1

    while index >= 0:
        group = groups[index]

        if group == 0:
            if result:
                zero_pending = True
        else:
            if result and (zero_pending or group < 1000):
                result += "零"

            result += convert_group(group) + large_units[index]
            zero_pending = False

        index -= 1

    return result


n = int(input("請輸入正整數："))
print(convert_number(n))
````

習題三：請勿使用for、range，而是利用iterable、iter()、next()、while、StopIteration來設計九九乘法表。

````python
numbers = iter(range(1, 10))

try:
    while True:
        a = next(numbers)

        multipliers = iter(range(1, 10))
        while True:
            try:
                b = next(multipliers)
                print(f"{a} × {b} = {a * b}", end="\t")
            except StopIteration:
                break

        print()

except StopIteration:
    pass
````
